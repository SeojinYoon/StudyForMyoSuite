```python
import opensim as osim

model = osim.Model("/mnt/sdb2/DeepProprioception/OpenSim/MOBL_ARMS_41.osim")

state = model.initSystem()

muscle = model.getMuscles().get("INFSP")

print("default length:", muscle.getLength(state))

print("optimal fiber length:", muscle.getOptimalFiberLength())

print("tendon slack length:", muscle.getTendonSlackLength())
```

```python
coordset = model.getCoordinateSet()

for i in range(coordset.getSize()):
    c = coordset.get(i)
    print(
        c.getName(),
        "default:", c.getDefaultValue(),
        "range:", c.getRangeMin(), c.getRangeMax(),
        "locked:", c.getLocked(state)
    )
```
