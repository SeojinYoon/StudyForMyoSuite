# Common Libraries

```python
import os, sys
import xml.etree.ElementTree as ET
```

# Custom Libraries

```python
sys.path.append("/home/seojin/Seojin_commonTool/Module/")
from XML.xml_util import parse_xml_with_includes
```

# Paths

```python
myo_arm_path = "/home/seojin/biomechanics_tools/myo_sim/arm/myoarm.xml"
myo_arm_full_path = "/mnt/ext1/seojin/temp/myoarm_full.xml"
myo_sim_dir_path = "/home/seojin/biomechanics_tools/myo_sim"
```

# Load xml

```python
root = parse_xml_with_includes(myo_arm_path)
```

# Save full xml

```python
tree = ET.ElementTree(root)
tree.write(myo_arm_full_path, encoding="utf-8", xml_declaration=True)
print(f"save: {myo_arm_full_path}")
```

# Change mesh path

```python
tree = ET.parse(myo_arm_full_path)
root = tree.getroot()
for elem in root.iter():
    for attr_name, attr_value in elem.attrib.items():
        if attr_value.startswith("../myo_sim/"):
            updated_value = attr_value.replace("../myo_sim/", f"{myo_sim_dir_path}/", 1)
            
            # 엘리먼트 속성값 업데이트
            elem.set(attr_name, updated_value)
            print(f"[{elem.tag}] {attr_name} 수정: {attr_value} -> {updated_value}")

tree = ET.ElementTree(root)
tree.write(myo_arm_full_path, encoding = "utf-8", xml_declaration = True)
print(f"save: {myo_arm_full_path}")
```
