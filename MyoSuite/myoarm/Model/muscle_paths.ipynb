# Common Libraries

```python
import os, sys, shutil
import numpy as np

if shutil.which("nvidia-smi") is not None:
    os.environ["MUJOCO_GL"] = "egl"
import mujoco
```

# Custom Libraries

```python
sys.path.append("/home/seojin/Seojin_commonTool/Module")

from sj_datastructure import make_ACS_timeseries
from sj_visualization import plot_3D_dataset
from sj_array import get_ACS_explicit_orientation
from mujoco_util import get_all_muscle_paths
```

# Params

```python
model_path = "/home/seojin/biomechanics_tools/myo_sim/arm/myoarm.xml"
coords = "RAI"
```

# Initialize model

```python
model = mujoco.MjModel.from_xml_path(model_path)


data = mujoco.MjData(model)
mujoco.mj_forward(model, data)
```

# Load muscle paths

```python
muscle_path_labels, muscle_locs = get_all_muscle_paths(model, data)
```

```python
muscle_path_ds = make_ACS_timeseries(np.expand_dims(muscle_locs, 0),
                                     labels = muscle_path_labels,
                                     coord_system = coords)
```

```python
target_muscles = ["DELT1", "DELT2", "DELT3"]
skeletons = []
for muscle in target_muscles:
    labels = [label for label in muscle_path_labels if label.startswith(muscle)]
    skeletons.append(list(zip(labels[:-1], labels[1:])))
skeletons = np.concatenate(skeletons, axis = 0)
```

```python
plot_3D_dataset(muscle_path_ds, visualize_coord_order = "LAI", skeletons = skeletons)
```
