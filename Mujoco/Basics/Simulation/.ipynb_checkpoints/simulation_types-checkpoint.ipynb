# Common Libraries

```python
import os
import numpy as np

os.environ.setdefault("PYOPENGL_PLATFORM", "egl")

import mujoco
```

# Custom Libraries

```python
def print_state(label):
    body_id = mujoco.mj_name2id(model, mujoco.mjtObj.mjOBJ_BODY, "pendulum")
    xpos = data.xpos[body_id].round(4)
    qacc = data.qacc.round(4)
    qpos = data.qpos.round(4)
    time = round(data.time, 4)
    print(f"[{label}]")
    print(f"  - time: {time}")
    print(f"  - qpos: {qpos}")
    print(f"  - body xpos: {xpos}")
    print(f"  - qacc: {qacc}\n")
```

# Initialize model

```python
xml = """
<mujoco>
  <option gravity="0 0 -9.81" timestep="0.01"/>
  <worldbody>
    <light pos="0 0 2"/>
    <body name="pendulum" pos="0 0 1">
      <joint name="pin" type="hinge" axis="0 1 0"/>
      <geom type="capsule" fromto="0 0 0 0 0 -0.5" size="0.04" mass="1.0"/>
    </body>
  </worldbody>
</mujoco>
"""

model = mujoco.MjModel.from_xml_string(xml)
data = mujoco.MjData(model)
```

# Result

```python
# Initial state: Lift the pendulum horizontally to 90 degrees
mujoco.mj_resetData(model, data)
data.qpos[0] = np.pi / 2.0

print("=== 0. 초기화 직후 (qpos만 변경, 아무 함수도 호출 안 함) ===")
print_state("초기 상태")

# 1. mj_kinematics 테스트
mujoco.mj_kinematics(model, data)
print("=== 1. mj_kinematics 호출 ===")
print_state("mj_kinematics")

# 2. mj_forward 테스트
mujoco.mj_forward(model, data)
print("=== 2. mj_forward 호출 ===")
print_state("mj_forward")

# 3. mj_step 테스트
mujoco.mj_step(model, data)
print("=== 3. mj_step 호출 ===")
print_state("mj_step")
```
